```python
# 6. Build Indic LMDB & Merged Character Dictionary (idempotent)
import os, io, glob, lmdb
import pyarrow.parquet as pq
from PIL import Image
from tqdm import tqdm

lmdb_out = '/content/AMSTR/IndicSTR12/indicstr12_lmdb'
merged_dict_path = 'tools/utils/mixed_latin_indic_dict.txt'

if not os.path.exists(lmdb_out) or not os.path.exists(merged_dict_path):
    os.makedirs(lmdb_out, exist_ok=True)
    env = lmdb.open(lmdb_out, map_size=1099511627776)

    parquet_files = sorted(glob.glob('/content/AMSTR/IndicSTR12/indicstr12_train/**/*.parquet', recursive=True))
    print(f'Packing {len(parquet_files)} Indic Parquet files into LMDB...')

    indic_chars = set()
    idx = 0

    with env.begin(write=True) as txn:
        for p_file in tqdm(parquet_files, desc='Converting to LMDB'):
            table = pq.read_table(p_file)
            pydict = table.to_pydict()
            images = pydict.get('image', [])
            texts = pydict.get('text', pydict.get('label', []))

            for img_obj, label in zip(images, texts):
                if not label or len(str(label).strip()) == 0:
                    continue

                if isinstance(img_obj, dict) and 'bytes' in img_obj and img_obj['bytes'] is not None:
                    img_bytes = img_obj['bytes']
                elif isinstance(img_obj, bytes):
                    img_bytes = img_obj
                else:
                    continue

                try:
                    img = Image.open(io.BytesIO(img_bytes))
                    w, h = img.size
                    if w * h == 0 or w < 8 or h < 8:
                        continue
                except Exception:
                    continue

                text_str = str(label).strip()
                indic_chars.update(list(text_str))

                idx += 1
                txn.put(f'image-{idx:09d}'.encode(), img_bytes)
                txn.put(f'label-{idx:09d}'.encode(), text_str.encode('utf-8'))
                txn.put(f'wh-{idx:09d}'.encode(), f'{w}_{h}'.encode('utf-8'))

        txn.put('num-samples'.encode(), str(idx).encode())

    env.close()
    print(f'✓ Created Indic LMDB with {idx:,} samples at {lmdb_out}')

    # Build merged Latin + Indic character dictionary
    with open('tools/utils/EN_symbol_dict.txt', 'r', encoding='utf-8') as f:
        latin_chars = [line.strip('\n').strip('\r') for line in f if line.strip()]

    all_chars = sorted(list(set(latin_chars).union(indic_chars)))
    with open(merged_dict_path, 'w', encoding='utf-8') as f:
        for ch in all_chars:
            f.write(f'{ch}\n')
    print(f'✓ Created merged character dictionary with {len(all_chars)} symbols at {merged_dict_path}')
else:
    print(f'✓ Indic LMDB and merged dictionary already exist at {lmdb_out}')
```

```python
# 7. Auto-detect, Auto-repair & Launch / Resume Training
import os, glob, shutil, torch

# ----------------------------------------------------------------------
# CHECKPOINT SELECTION:
# 'latest': Resumes from step 107k+ (continue through Epoch 3)
# 'best'  : Resumes from highest recorded accuracy checkpoint (70.74%)
# ----------------------------------------------------------------------
RESUME_PREFERENCE = 'latest'

checkpoint_dir = '/content/drive/MyDrive/svtrv2x_checkpoints/mixed_latin_indic_tiny'
os.makedirs(checkpoint_dir, exist_ok=True)

# Auto-collect checkpoints from Google Drive or /content/
search_locations = [
    checkpoint_dir,
    '/content',
    '/content/drive/MyDrive',
    '/content/AMSTR/output'
]

for loc in search_locations:
    if os.path.exists(loc) and loc != checkpoint_dir:
        for f in os.listdir(loc):
            if f.endswith('.pth') and any(tag in f for tag in ['latest', 'best', 'step', 'epoch', 'recovered']):
                dst = os.path.join(checkpoint_dir, f)
                if not os.path.exists(dst):
                    print(f'Copying found checkpoint {f} to {checkpoint_dir}...')
                    shutil.copy2(os.path.join(loc, f), dst)

# Inspect and auto-repair checkpoints
valid_checkpoints = {}
all_ckpts = [f for f in os.listdir(checkpoint_dir) if f.endswith('.pth')]

print('=' * 70)
print(f'Inspecting checkpoints in: {checkpoint_dir}')
print('=' * 70)

for cand in sorted(all_ckpts):
    cand_path = os.path.join(checkpoint_dir, cand)
    try:
        ckpt = torch.load(cand_path, map_location='cpu')
        state_dict = ckpt.get('state_dict', {})
        if len(state_dict) == 0:
            continue

        # Auto-repair BatchNorm tracking buffers if any NaNs exist
        repaired = False
        for name, tensor in state_dict.items():
            if torch.isnan(tensor).any() or torch.isinf(tensor).any():
                if 'running_mean' in name:
                    tensor.zero_()
                    repaired = True
                elif 'running_var' in name:
                    tensor.fill_(1.0)
                    repaired = True
                elif 'num_batches_tracked' in name:
                    tensor.zero_()
                    repaired = True
                else:
                    tensor[torch.isnan(tensor) | torch.isinf(tensor)] = 0.0
                    repaired = True

        if repaired:
            torch.save(ckpt, cand_path)
            print(f'✓ Auto-repaired BatchNorm running statistics in {cand}')

        has_nan = any(torch.isnan(v).any() for v in state_dict.values())
        step = ckpt.get('global_step', 0)
        epoch = ckpt.get('epoch', 0)
        metric = ckpt.get('metrics', {})
        acc = metric.get('acc', 0.0)

        if not has_nan:
            valid_checkpoints[cand] = {
                'path': cand_path,
                'step': step,
                'epoch': epoch,
                'acc': acc,
                'metric': metric
            }
            print(f'  ✓ {cand:<16} | Epoch: {epoch} | Step: {step:<7} | Val Acc: {acc*100:.2f}%')
        else:
            print(f'  ✗ {cand:<16} | Corrupted with unrecoverable NaNs')
    except Exception as e:
        print(f'  ⚠ Could not inspect {cand}: {e}')

# Determine which checkpoint to resume from
resume_checkpoint = None
resume_step = 0

if RESUME_PREFERENCE == 'latest' and 'latest.pth' in valid_checkpoints:
    resume_checkpoint = valid_checkpoints['latest.pth']['path']
    resume_step = valid_checkpoints['latest.pth']['step']
elif RESUME_PREFERENCE == 'best' and 'best.pth' in valid_checkpoints:
    resume_checkpoint = valid_checkpoints['best.pth']['path']
    resume_step = valid_checkpoints['best.pth']['step']
elif len(valid_checkpoints) > 0:
    # Fallback to the checkpoint with the highest step count
    highest = max(valid_checkpoints.values(), key=lambda x: x['step'])
    resume_checkpoint = highest['path']
    resume_step = highest['step']

config_path = 'configs/rec/svtrv2/svtrv2x_tiny_mixed_latin_indic.yml'
cmd = f'python tools/train_rec.py -c {config_path}'

if resume_checkpoint is not None:
    print(f'\n>>> Resuming training from: {resume_checkpoint} (Step {resume_step})')
    cmd += f' -o Global.checkpoints={resume_checkpoint}'
else:
    print('\n>>> No clean checkpoint found on Google Drive. Training from scratch.')

print(f'Running command:\n{cmd}\n')
!{cmd}
```
